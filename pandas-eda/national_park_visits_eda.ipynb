# Exploring U.S. National Park Visits

This notebook uses annual recreation visit counts for U.S. national parks from 1979 to 2024. I will inspect the data and use a line chart to compare the yearly visit counts for the five most visited parks in 2024.

The dataset is from [Responsible Datasets in Context](https://www.responsible-datasets-in-context.com/posts/np-data/). The project explains that park visit counts are estimates and that counting methods can differ between parks.


## 1. Load the dataset

Pandas reads the CSV file from the dataset's public source and stores it in a DataFrame called `parks`.


```python
import pandas as pd
import altair as alt

url = "https://raw.githubusercontent.com/melaniewalsh/responsible-datasets-in-context/main/datasets/national-parks/US-National-Parks_RecreationVisits_1979-2024.csv"
parks = pd.read_csv(url)
```

## 4. Count rows for each park

`value_counts()` shows how many yearly records appear for each park. The result is sorted from most rows to fewest.


```python
parks["ParkName"].value_counts().head(10)
```

## 5. Find the five most visited parks in 2024

First, I filter the DataFrame to 2024. Then I sort by recreation visits and keep the five parks with the largest counts.


```python
visits_2024 = parks[parks["Year"] == 2024]
top_parks = visits_2024.sort_values(
    by="RecreationVisits", ascending=False
).head(5)

top_parks[["ParkName", "RecreationVisits"]]
```

## 6. Compare annual visits with an Altair line chart

I use the five parks selected from 2024 and keep their records for every year in the dataset. In the chart, year is on the x-axis, recreation visits are on the y-axis, and color distinguishes the parks. Hover over a point to see its park, year, and visit count. The chart also supports zooming and panning.


```python
chart_data = parks[parks["ParkName"].isin(top_parks["ParkName"])]

chart = alt.Chart(chart_data).mark_line(point=True).encode(
    x=alt.X("Year:Q", title="Year", axis=alt.Axis(format="d")),
    y=alt.Y("RecreationVisits:Q", title="Recreation visits"),
    color=alt.Color("ParkName:N", title="Park"),
    tooltip=[
        alt.Tooltip("ParkName:N", title="Park"),
        alt.Tooltip("Year:Q", title="Year", format="d"),
        alt.Tooltip("RecreationVisits:Q", title="Visits", format=",")
    ]
).properties(
    title="Annual Visits for the Five Most Visited Parks in 2024",
    width=700,
    height=400
)

chart.interactive()
```

## What I noticed

The dataset has **2,846 rows and 5 columns**, with no missing values. Each row records one park's estimated recreation visits for one year. Great Smoky Mountains National Park had the highest count among these parks in 2024, with **12,191,834** visits.

The chart shows that the five parks have different levels of visitation. All five had fewer recorded visits in 2020 than in 2019, followed by a rise in 2021. The size of the change differs by park. These values count estimated recreation visits, not unique people, and the dataset notes that counting methods are not identical at every park.
