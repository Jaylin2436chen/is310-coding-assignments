After exploring the U.S National Park Visit Data between 1979-2024, I decided to compare the five parks that contained the highest annual visits within 2024. 

```python
visits_2024 = df[df["Year"] == 2024]

top_parks = visits_2024.sort_values(
    "RecreationVisits", ascending=False).head(5)

chart_data = df[df["ParkName"].isin(top_parks["ParkName"])]

alt.Chart(chart_data).mark_line(point=True).encode(
    x=alt.X("Year:Q", title="Year"),
    y=alt.Y("RecreationVisits:Q", title="Number of Visits"),
    color=alt.Color("ParkName:N", title="Park"),
    tooltip=["ParkName:N", "Year:Q", "RecreationVisits:Q"]).properties(
    title="Annual Visits for the Top Five Most Visited Parks in 2024",
    width=700,
    height=400
).interactive()

```

The visualization I created allows users to scroll through each year and park to see a percise number of annual visits per year. As I created the altair visualization, I saw that there was a steady growth of visits per year within the top five parks all the way up until 2020-2021. This is most likely caused by the pandemic (Covid-19) that happened causing everyone to go into quarantine. Right after it however, we can see that the growth of national park visits begin to rise again.
